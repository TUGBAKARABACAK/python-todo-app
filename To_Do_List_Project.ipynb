```python
import os

# --- Sabitler ---
DOSYA_ADI = "gorev_kaydi.txt"
KODLAMA = "utf-8"
GOREV_LISTESI = []  # Görevleri tutacak ana liste

# --- Yardımcı Fonksiyonlar ---
def temiz_metin_al(mesaj):
    """Kullanıcıdan boş olmayan bir metin girişi alır."""
    while True:
        giris = input(mesaj).strip()
        if giris:
            return giris
        else:
            print("❌ Giriş boş bırakılamaz.")

def gecerli_gorev_no_al(islem_tipi):
    """Kullanıcıdan geçerli (1 tabanlı) görev numarası alır."""
    while True:
        try:
            no_str = input(f"{islem_tipi.capitalize()}mek istediğiniz görevin Numarası: ").strip()
            gorev_no = int(no_str)

            if 1 <= gorev_no <= len(GOREV_LISTESI):
                return gorev_no - 1
            else:
                print("❌ Geçersiz görev numarası! Lütfen listedeki geçerli bir numarayı girin.")
                return None

        except ValueError:
            print("❌ Hatalı giriş! Lütfen görev numarası olarak **bir sayı** girin.")
        except Exception as e:
            print(f"❌ Beklenmedik bir hata oluştu: {e}")
            return None
```

```python
# --- Dosya İşlemleri Fonksiyonları ---

def gorevleri_yukle():
    """Görevleri dosyadan okur ve GOREV_LISTESI'ne yükler."""
    global GOREV_LISTESI
    try:
        with open(DOSYA_ADI, 'r', encoding=KODLAMA) as f:
            GOREV_LISTESI = [satir.strip() for satir in f if satir.strip()]
        print("✅ Görevler başarıyla yüklendi.")
    except FileNotFoundError:
        GOREV_LISTESI = []
        print("ℹ️ Kayıt dosyası bulunamadı. Yeni bir boş görev listesi oluşturuldu.")
    except IOError as e:
        print(f"❌ Dosya okuma hatası oluştu: {e}")
        GOREV_LISTESI = []
    return GOREV_LISTESI

def gorevleri_kaydet():
    """GOREV_LISTESI'ni dosyaya yazar."""
    try:
        with open(DOSYA_ADI, 'w', encoding=KODLAMA) as f:
            for gorev in GOREV_LISTESI:
                f.write(gorev + '\n')
        print("💾 Değişiklikler başarıyla kaydedildi.")
        return True
    except IOError as e:
        print(f"❌ Dosyaya yazma hatası oluştu: {e}")
        return False
```

```python
# --- CRUD: Görüntüleme (Oku) ve Ekleme (Oluştur) ---

def gorevleri_goster():
    """Mevcut görevleri numaralandırılmış şekilde listeler."""
    print("\n--- MEVCUT GÖREVLER ---")
    if not GOREV_LISTESI:
        print("🎉 Listenizde tamamlanması beklenen herhangi bir görev yok!")
    else:
        for i, gorev in enumerate(GOREV_LISTESI, 1):
            print(f"[{i:02d}] {gorev}")
    print("------------------------")

def gorev_ekle():
    """Kullanıcıdan yeni bir görev alır ve listeye ekler."""
    print("\n--- YENİ GÖREV EKLEME ---")
    yeni_gorev = temiz_metin_al("Lütfen eklemek istediğiniz görevi yazın: ")

    if yeni_gorev:
        GOREV_LISTESI.append(yeni_gorev)
        print(f"✅ Görev: **'{yeni_gorev}'** listeye başarıyla eklendi.")
        gorevleri_kaydet()
```

```python
# --- CRUD: Güncelleme ve Silme ---

def gorev_guncelle():
    """Seçilen görevi kullanıcıdan alınan yeni metinle günceller."""
    if not GOREV_LISTESI:
        print("⚠️ Düzenlenecek bir görev bulunmuyor. Lütfen önce görev ekleyin.")
        return

    gorevleri_goster()

    index = gecerli_gorev_no_al("Düzenle")

    if index is not None:
        eski_gorev = GOREV_LISTESI[index]
        print(f"\n--- GÖREV DÜZENLEME (Görev #{index + 1}) ---")

        yeni_metin = temiz_metin_al(f"Yeni görev metnini girin (Eski: '{eski_gorev}'): ")

        if yeni_metin:
            GOREV_LISTESI[index] = yeni_metin
            print(f"✅ Görev **{index + 1}** başarıyla güncellendi.")
            gorevleri_kaydet()

def gorev_sil():
    """Seçilen görevi listeden siler."""
    if not GOREV_LISTESI:
        print("⚠️ Silinecek bir görev bulunmuyor.")
        return

    gorevleri_goster()

    index = gecerli_gorev_no_al("Sil")

    if index is not None:
        silinen_gorev = GOREV_LISTESI.pop(index)
        print(f"✅ Görev: **'{silinen_gorev}'** listeden kalıcı olarak silindi.")
        gorevleri_kaydet()
```

```python
# --- Ana Uygulama Akışı ve Başlatıcı ---

def menuyu_goster():
    """Ana menüyü gösterir ve kullanıcıdan seçim alır."""
    print("\n==================================")
    print("📋 PYTHON GÖREV YÖNETİM UYGULAMASI")
    print("==================================")
    print("1. Tüm Görevleri Görüntüle")
    print("2. Yeni Görev Oluştur")
    print("3. Bir Görevi Güncelle")
    print("4. Bir Görevi Sil")
    print("5. Uygulamadan Çık")
    print("----------------------------------")

    secim = input("Lütfen yapmak istediğiniz işlemin numarasını girin (1-5): ").strip()
    return secim

def baslat():
    """Uygulamanın ana döngüsünü çalıştırır."""

    print("👋 Uygulama Başlatılıyor...")
    gorevleri_yukle()

    calisiyor = True
    while calisiyor:
        secim = menuyu_goster()

        if secim == '1':
            gorevleri_goster()
        elif secim == '2':
            gorev_ekle()
        elif secim == '3':
            gorev_guncelle()
        elif secim == '4':
            gorev_sil()
        elif secim == '5':
            print("🚪 Program sonlandırılıyor. İyi çalışmalar dilerim!")
            calisiyor = False
        else:
            print("❌ Geçersiz seçim! Lütfen menüdeki numaralardan birini (1, 2, 3, 4, 5) girin.")

# Uygulamayı doğrudan başlat
if __name__ == "__main__":
    baslat()
```
